```python
#importing libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt



```

```python
#making dataframe

data={
    'name':['paras','ishant','pratyaya','ujjwal','garv','shivansh','ashrey'],
    'age':[19,17,28,35,23,18,22],
    'roll_no':[1,2,3,4,5,6,7],
    'course':['b.tech','bba','b.pharma','b.tech','bba','b.pharma','b.tech'],
    'marks':[99,98,96,84,89,90,87]
}

data=pd.DataFrame(data)
data.to_csv('data1.csv')
```

```python
#converting dataframe to csv

data=pd.DataFrame(data)
data.to_csv('data1.csv')
```

```python
#reading csv file

df=pd.read_csv('data1.csv')
data
```

```python
#plotting scatter plot


plt.scatter(df['roll_no'],df['age'])
plt.xlabel('roll_no')
plt.ylabel('age')
plt.title('roll_no vs age')
plt.show()
```

```python
#printng y coloumn in scatter plot

x=df['age']
x
```

```python
#printng x coloumn in scatter plot


y=df['roll_no']
y
```

```python
#reshaping coloumn 1

X=df['age']
X=np.array(X).reshape(-1,1)
X
```

```python
#reshaping coloumn 2

Y=df['roll_no']
Y=np.array(y).reshape(-1,1)
Y
```

```python
#importing libraries

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
#training and testing data

X_train,X_test,Y_train,y_test = train_test_split(X,Y,test_size=0.2)
X_train,X_test,Y_train,y_test
```

```python
#training data

model=LinearRegression()
model.fit(X_train,Y_train)
```

```python
Y_pred = model.predict(X_test)
Y_pred

```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score


print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred, average='weighted'):.4f}")
print(f"Recall: {recall_score(y_test, y_pred, average='weighted'):.4f}")
print(f"F1-score: {f1_score(y_test, y_pred, average='weighted'):.4f}")
```
